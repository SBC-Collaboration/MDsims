```python
import sys
import sqlite3
import pandas as pd



print("Python executable:", sys.executable)
print("SQLite version:", sqlite3.sqlite_version)
```

```python
cx = sqlite3.connect("test_db.db")
cu = cx.cursor()
```

```python
cu.execute("create table if not exists MDSims(Run_ID, n_cells, Therm_kT, density)")
```

```python
##Cell to make fake data
from datetime import datetime, timezone

def create_run_id():
    """Return a UTC Run ID formatted as YYYYMMDDHHMMSS."""
    return datetime.now(timezone.utc).strftime("%Y%m%d%H%M%S")



run_id = create_run_id()
ncell = 45
kT =  .9
rho = .5

print("Run ID:", run_id)
print("Ncells:", ncell)
print("kT:", kT)
print("Density:", rho)
```

```python
cu.execute("insert into MDSims values (?, ?, ?, ?)", (run_id, ncell, kT, rho))
```
