```python
import matplotlib.pyplot as plt
from md_Helpers import open_run

run_ids = [
  "20260909210822",
  "20260909210949",
  "20260909211115",
  "20260909202117",
  "20260909210117",
  "20260909210328",
  "20260909211334",
  "20260909211645",
  "20260909211957",
  "20260909201254",
  "20260909212511",
  "20260909213227"
]

fig, ax = plt.subplots(figsize=(9, 5))

for run_id in run_ids:
    run = open_run(run_id)
    logs = run.logs_dataframe()

    ax.plot(
        logs["this_lj_time"],
        logs["pressure"],
        label=run_id,
    )
ax.set_xlabel("LJ time within run")
ax.set_ylabel("Potential energy per particle, PE/N")
ax.set_title("Potential energy per particle comparison")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```
