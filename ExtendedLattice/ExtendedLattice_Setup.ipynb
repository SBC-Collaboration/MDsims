```python
import pandas as pd
from IPython.display import display

from md_Helpers import (
    ExpandedFCCConfig,
    ProjectPaths,
    open_run,
    run_expanded_fcc,
)

paths = ProjectPaths()

results = []

for density in [0.68]:
    config = ExpandedFCCConfig(
        n_fcc_cells=30,
        target_rho=density,
        nsteps=1_000_000,
        kT=0.9,
        seed=1,
        dt=0.002,
        epsilon_LJ=1.0,
        sigma_LJ=1.0,
        r_cut_LJ=2.5,
        r_on_LJ=2.0,
        buffer_LJ=0.4,
        lj_mode="xplor",

        # Expanded-FCC controls:
        center_length_scale=3.0,
        side_extension=2.0,          # Adds L on both sides → total length 3L
        side_density_divisor=25.0,    # Side density = target_rho / num
        com_recenter_period=10_000,  # Recenter and save every 10k steps

        progress_period=10_000,
        #notes="extended center",
    )

    result = run_expanded_fcc(config, project_paths=paths)
    results.append({
        "target_density": density,
        "run_id": result["run_id"],
        "status": result["status"],
        "skipped": result["skipped"],
        "particles_per_side": result.get("particles_per_side"),
        "side_density": result.get("side_density"),
        "num_frames": result.get("num_frames"),
    })
```
