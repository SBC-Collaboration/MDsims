```python
import sys
import sqlite3
import pandas as pd
from Database import *


print("Python executable:", sys.executable)
print("SQLite version:", sqlite3.sqlite_version)
```

```python
cx = sqlite3.connect("MDSims.db")
cu = cx.cursor()
```

```python
MasterTableRows = [
    "Run_ID",
    "Sim_Type",
    "N_Cells", 
    "Nsteps", 
    "Current_Nstep",
    "ElapsedTime", 
    "StartTime", 
    "EndTime", 
    "Last_Update_Time",  
    "Status", 
    "Stop_Reason", 
    "Status_Message", 
    "Notes",
    "Run_Signature"
]

MasterTableRows = [
    "Run_ID",
    "Sim_Type",
    "N_Cells", 
    "Nsteps", 
    "Run_Signature"
]

MasterConstruct = ', '.join(MasterTableRows)


cu.execute("create table if not exists MD_Master("+MasterConstruct+")")
```

```python
##Cell to make fake data

ncell = 30
kT =  .9
rho = .5
sim_type = 'Thermalization'
nsteps = 10_000

run_signature = create_run_signature(ncell, kT, rho)
run_id = create_run_id()


print("Run ID:", run_id)
print("Sim Type:", sim_type)
print("Ncells:", ncell)
print("Nsteps:", nsteps)
print("Density:", rho)

cu.execute("insert into MD_Master values (?, ?, ?, ?, ?)", (run_id, None, ncell, nsteps, run_signature))
```
