```python
import matplotlib.pyplot as plt
from md_Helpers import open_run

run_ids = [
    "20260903215610",
    "20260903214936",
    "20260903221748",
]

fig, ax = plt.subplots(figsize=(9, 5))

for run_id in run_ids:
    run = open_run(run_id)
    logs = run.logs_dataframe()

    ax.plot(
        logs["this_lj_time"],
        logs["kinetic_temperature"],
        label=run_id,
    )

ax.set_xlabel("LJ time within run")
ax.set_ylabel("Potential energy per particle, PE/N")
ax.set_title("Potential energy per particle comparison")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```
