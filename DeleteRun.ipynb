```python
import pandas as pd
from IPython.display import display
from md_Helpers import ProjectPaths, SQLiteRunDatabase, master_dataframe

database = SQLiteRunDatabase(ProjectPaths().database)
database.initialize()

failed_runs = master_dataframe(database)
failed_runs = (
    failed_runs.loc[failed_runs["Status"].eq("Cancelled")]
    .drop(
        columns=[
            "Run_Signature",
            "StartTime",
            "EndTime",
            "Last_Update_Time",
        ],
        errors="ignore",
    )
    .reset_index(drop=True)
)

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None,
    "display.max_colwidth", None,
):
    display(failed_runs.style.hide(axis="index"))

print(f"{len(failed_runs)} failed run(s)")
```
