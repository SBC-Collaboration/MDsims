```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

from md_Helpers import cavitation as cav
from md_Helpers import visualization as vh
from md_Helpers import seitz

eos_csv = Path(
    "/exp/e961/data/MDsims-data/pnichols/Master_CSVs_v3/"
    "thermalization_master_ncells_30.csv"
)

starting_radii = [3, 3.5, 4, 4.5, 5, 5.5, 6, 6.5, 7]

results = []

for radius in starting_radii:
    print(f"Running/checking starting radius {radius}...")

    evolution = cav.get_or_create_cavitation(
        n_fcc_cells=30,
        target_rho=0.71,
        kT=0.80,
        source_nsteps=1_000_000,
        radius=radius,
        evolve_nsteps=100_000,
        create_source_if_missing=True,
        overwrite_source=False,
        overwrite=False,
    )

    check = vh.fit_and_animate_final_bubble(
        evolution,
        nbins=12,
        nframes=5,
        skip=5,
        tail_fraction=0.2,
    )

    seitz_result = seitz.seitz_threshold_from_files(
        metadata_path=evolution["paths"]["log_path"],
        eos_table=eos_csv,
        radius=check["bubble_radius"],
    )

    results.append({
        "starting_radius": radius,
        "final_radius": check["bubble_radius"],
        "final_volume": check["bubble_volume"],
        "volume_fraction": check["bubble_volume_fraction"],
        "status": evolution.get("status"),
        "p0": seitz_result["p0"],
        "rho0": seitz_result["rho0"],
        "u0": seitz_result["u0"],
        "n_cavity": seitz_result["n_cavity"],
        "u_cavity": seitz_result["u_cavity"],
        "q_seitz": seitz_result["q_seitz"],
        "log_path": str(evolution["paths"]["log_path"]),
    })

radius_df = pd.DataFrame(results)
display(radius_df)

plt.figure(figsize=(6, 4))
plt.plot(radius_df["final_radius"], radius_df["q_seitz"], marker="o")
plt.xlabel("Final fitted radius")
plt.ylabel(r"$Q_{\mathrm{Seitz}}$")
plt.title("Seitz threshold vs final bubble radius, kT=0.80, rho=0.71")
plt.grid(alpha=0.3)
plt.show()
```
