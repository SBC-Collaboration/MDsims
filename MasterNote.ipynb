```python
from datetime import datetime, timezone
from md_Helpers import open_run

#NOTE = "Write your new note here."
NOTE = "log_period=1000"


run = open_run(RUN_ID)
current_notes = run.database.get_run(RUN_ID)["Notes"] or ""

timestamp = datetime.now(timezone.utc).strftime("%Y-%m-%d %H:%M:%S UTC")
note_entry = f"[{timestamp}] {NOTE}"

updated_notes = (
    f"{current_notes}\n{note_entry}"
    if current_notes
    else note_entry
)

run.database.update_master(RUN_ID, Notes=updated_notes)

print(f"Note added to Run_ID {RUN_ID}.")
```
