# V3 Saved-File Migration

This notebook reorganizes existing V2 saved files into the V3 saving scheme without rerunning simulations.

Start with `DRY_RUN = True`. Review the returned table. Then set `DRY_RUN = False` when you are ready to create the V3 files.

```python
import importlib
import pandas as pd

import md_Helpers.migration as migration

importlib.reload(migration)

# ============================================================
# Migration controls
# ============================================================

# Keep this True until you have reviewed the plan.
DRY_RUN = True

# V3 migration is copy-only so old V2 files stay unchanged.
LINK_MODE = "copy"

# Adds /metadata/state, /metadata/run, /metadata/lj, etc. to migrated logs.
# Original flat /metadata.attrs are preserved in the copied HDF5 logs.
WRITE_V3_METADATA = True

# Keep False unless you intentionally want to replace existing V3 files.
OVERWRITE = False

# Use a small integer while testing, then None for the full migration.
LIMIT = 10

plan = migration.migrate_v2_to_v3(
    dry_run=DRY_RUN,
    link_mode=LINK_MODE,
    overwrite=OVERWRITE,
    migrate_lattices=True,
    migrate_thermalized=True,
    migrate_cavitation_states=False,
    write_v3_metadata=WRITE_V3_METADATA,
    limit=LIMIT,
)

print("Rows in migration plan:", len(plan))

if len(plan) > 0:
    display(plan["action"].value_counts(dropna=False))
    display(plan.head(20))
else:
    display(plan)
```
