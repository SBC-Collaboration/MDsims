# V4 Thermalization Starter
This notebook uses the local SQLite backend and the output root configured in `md_Helpers/paths.py`.

```python
from md_Helpers import (
    ProjectPaths,
    SQLiteRunDatabase,
    ThermalizationConfig,
    display_thermalization_table,
    run_thermalization,
)

paths = ProjectPaths()
print("Output root:", paths.top_directory)
print("Local database:", paths.database)
```

```python
config = ThermalizationConfig(
    n_fcc_cells=30,
    target_rho=0.6,
    nsteps=500_000,
    kT=0.9,
    log_period=1_000,
    seed=1,
    dt=0.005,
    epsilon_LJ=1.0,
    sigma_LJ=1.0,
    r_cut_LJ=2.5,
    buffer_LJ=0.4,
    lj_mode="xplor",
    r_on_LJ=2.0,
)

print("Run signature:", config.run_signature)
```

```python
result = run_thermalization(config, project_paths=paths)
result
```

```python
database = SQLiteRunDatabase(paths.database)
database.query_runs(Sim_Type="Thermalization")
```

```python
# No filters displays the entire Thermalization table.
display_thermalization_table(database)
```

```python
# Scalars are exact matches, tuples are inclusive ranges, and lists
# contain any acceptable value. Set limit=None to return every match.
matches = display_thermalization_table(
    database,
    limit=100,
    Therm_kT=(0.85, 0.95),
    Density_End=(0.5, 0.65),
    Nsteps=[10_000, 100_000, 500_000, 1_000_000],
    Phase_Separation_Status="Separated",
)
matches
```
