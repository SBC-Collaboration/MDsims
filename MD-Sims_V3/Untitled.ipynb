```python
from importlib import reload
from pathlib import Path
import shlex

import h5py
import numpy as np
import pandas as pd

from md_Helpers import (
    excitation_evolution,
    hot_spike,
    master_csv,
    run_logs,
)

reload(run_logs)
reload(excitation_evolution)
reload(hot_spike)
reload(master_csv)

# Start one persistent log for the entire rerun sweep.
run_log_path = run_logs.configure_run_logging(
    enabled=True,
    notebook_name="Excitation_Failed_Reruns",
)

progress_print = run_logs.progress_print
tail_command = f"tail -f {shlex.quote(str(run_log_path))}"

print("Run log:")
print(run_log_path)
print("\nMonitor from a terminal with:")
print(tail_command)

# Select unfinished runs, but do not rerun valid terminal bubbles.
files_complete = (
    excitation_master["files_complete"]
    .fillna(False)
    .astype(bool)
)

terminal_bubble = (
    excitation_master["outcome"]
    .fillna("")
    .eq("bubble")
)

rerun_table = excitation_master.loc[
    ~files_complete
    & ~terminal_bubble
    & excitation_master["status"].isin([
        "running",
        "segment_1_complete",
        "failed_segment_1",
        "failed_segment_2",
        "summary_failed",
    ])
].copy()

progress_print("=" * 80)
progress_print(
    f"Found {len(rerun_table)} excitation evolutions to overwrite."
)
progress_print(f"Run log: {run_log_path}")
progress_print("=" * 80)

display(
    rerun_table[
        [
            "n_fcc_cells",
            "energy_deposition",
            "ensemble",
            "status",
            "run_path",
        ]
    ]
)
```

```python
def read_attrs(hdf5_path, group_path):
    hdf5_path = Path(hdf5_path)

    if not hdf5_path.exists():
        return {}

    try:
        with h5py.File(hdf5_path, "r") as hdf:
            if group_path not in hdf:
                return {}

            return {
                key: (
                    value.item()
                    if isinstance(value, np.generic)
                    else value
                )
                for key, value in hdf[group_path].attrs.items()
            }
    except Exception:
        return {}


def usable(value):
    if value is None:
        return False

    try:
        return not pd.isna(value)
    except (TypeError, ValueError):
        return True


def first_valid(*values):
    for value in values:
        if usable(value):
            return value
    return None


def path_value(path, prefix, converter=str):
    for part in Path(path).parts:
        if part.startswith(prefix):
            return converter(part[len(prefix):])
    return None


def path_float(path, prefix, suffix=""):
    for part in Path(path).parts:
        if not part.startswith(prefix):
            continue

        value = part[len(prefix):]

        if suffix and value.endswith(suffix):
            value = value[:-len(suffix)]

        try:
            return float(value)
        except ValueError:
            continue

    return None


def infer_base_folder(manifest_path):
    parts = Path(manifest_path).parts

    if "FCC" not in parts:
        raise ValueError(
            f"Cannot identify dataset root from {manifest_path}"
        )

    return Path(*parts[:parts.index("FCC")])


def optional_float(value):
    return float(value) if usable(value) else None


print("Parameter-recovery helpers are ready.")
```

```python
rerun_results = []
total_reruns = len(rerun_table)

for rerun_number, (index, row) in enumerate(
    rerun_table.iterrows(),
    start=1,
):
    run_path = Path(row["run_path"])
    manifest_path = run_path / "evolution_manifest.hdf5"

    record = {
        "master_index": index,
        "manifest_path": str(manifest_path),
        "old_status": row["status"],
        "rerun_status": "not_started",
        "result_status": "",
        "outcome": "",
        "error": "",
    }

    progress_print("")
    progress_print("=" * 80)
    progress_print(
        f"Preparing rerun {rerun_number}/{total_reruns}; "
        f"master row={index}; old status={row['status']}"
    )
    progress_print(f"Manifest: {manifest_path}")

    try:
        if not manifest_path.exists():
            raise FileNotFoundError(
                f"Missing manifest: {manifest_path}"
            )

        state = read_attrs(
            manifest_path,
            "metadata/state",
        )
        run = read_attrs(
            manifest_path,
            "metadata/run",
        )
        source = read_attrs(
            manifest_path,
            "metadata/source",
        )
        creation = read_attrs(
            manifest_path,
            "metadata/creation",
        )
        segment_1 = read_attrs(
            manifest_path,
            "metadata/segments/segment_1",
        )
        segment_2 = read_attrs(
            manifest_path,
            "metadata/segments/segment_2",
        )

        n_fcc_cells = int(first_valid(
            state.get("n_fcc_cells"),
            row.get("n_fcc_cells"),
            path_value(run_path, "n_cells_", int),
        ))

        target_rho = float(first_valid(
            source.get("source_rho"),
            state.get("source_rho"),
            row.get("density"),
            path_value(run_path, "rho_", float),
        ))

        kT = float(first_valid(
            source.get("source_kT"),
            state.get("kT"),
            row.get("temp"),
            path_value(run_path, "source_kT_", float),
        ))

        source_nsteps = int(first_valid(
            source.get("source_nsteps"),
            row.get("source_nsteps"),
            path_value(run_path, "source_nsteps_", int),
        ))

        source_seed = int(first_valid(
            source.get("source_seed"),
            row.get("source_seed"),
            path_value(run_path, "source_seed_", int),
        ))

        radius = float(first_valid(
            creation.get("radius"),
            row.get("radius"),
            path_value(run_path, "radius_", float),
        ))

        injected_energy = float(first_valid(
            creation.get("requested_injected_energy"),
            row.get("energy_deposition"),
            path_value(run_path, "energy_", float),
        ))

        method = str(first_valid(
            creation.get("energy_dump_method"),
            row.get("excitation_method"),
            path_value(run_path, "method_", str),
        ))

        dt1 = float(first_valid(
            run.get("dt1"),
            segment_1.get("dt"),
            row.get("dt1"),
        ))

        nsteps1 = int(first_valid(
            run.get("nsteps1"),
            segment_1.get("nsteps"),
            row.get("nsteps1"),
        ))

        dt2 = float(first_valid(
            run.get("dt2"),
            segment_2.get("dt"),
            row.get("dt2"),
        ))

        nsteps2 = int(first_valid(
            run.get("nsteps2"),
            segment_2.get("nsteps"),
            row.get("nsteps2"),
        ))

        evolve_seed = int(first_valid(
            run.get("seed"),
            row.get("evolve_seed"),
        ))

        ensemble = str(first_valid(
            run.get("ensemble"),
            row.get("ensemble"),
            "NVE",
        )).upper()

        pressure = optional_float(
            run.get("pressure")
        )

        tauS = optional_float(first_valid(
            run.get("tauS"),
            row.get("tauS"),
            path_float(run_path, "tauS_"),
        ))

        pressure_couple = str(first_valid(
            run.get("pressure_couple"),
            path_value(run_path, "couple_", str),
            "xyz",
        ))

        barostat_gamma = float(first_valid(
            run.get("barostat_gamma"),
            path_float(run_path, "gamma_"),
            0.0,
        ))

        outer_mask_fraction = optional_float(first_valid(
            run.get("outer_mask_diameter_fraction"),
            path_float(
                run_path,
                "outer_mask_diameter_",
                suffix="L",
            ),
            0.75,
        ))

        nph_mask_controls_box = bool(first_valid(
            run.get("nph_mask_controls_box"),
            (
                "pressure_control_outer_mask_experimental"
                in run_path.parts
            ),
            False,
        ))

        lower_bound = run.get(
            "nph_min_volume_ratio"
        )
        upper_bound = run.get(
            "nph_max_volume_ratio"
        )

        if usable(lower_bound) and usable(upper_bound):
            volume_bounds = (
                float(lower_bound),
                float(upper_bound),
            )
        elif ensemble == "NPH":
            volume_bounds = (0.75, 1.5)
        else:
            volume_bounds = None

        safety_period = int(first_valid(
            run.get("nph_safety_check_period"),
            100,
        ))

        random_location = bool(first_valid(
            creation.get("random_location"),
            any(
                part.startswith("random_center")
                for part in run_path.parts
            ),
            False,
        ))

        base_folder = infer_base_folder(
            manifest_path
        )

        progress_print(
            f"Starting rerun {rerun_number}/{total_reruns}: "
            f"ncells={n_fcc_cells}, rho={target_rho}, "
            f"kT={kT}, radius={radius}, "
            f"energy={injected_energy}, ensemble={ensemble}"
        )
        progress_print(
            f"Segment 1: dt={dt1}, nsteps={nsteps1}; "
            f"segment 2: dt={dt2}, nsteps={nsteps2}"
        )
        progress_print(
            f"pressure={pressure}, tauS={tauS}, "
            f"volume_bounds={volume_bounds}"
        )

        result = hot_spike.get_or_create_hot_spike(
            n_fcc_cells=n_fcc_cells,
            target_rho=target_rho,
            kT=kT,
            source_nsteps=source_nsteps,
            source_seed=source_seed,

            radius=radius,
            injected_energy=injected_energy,
            method=method,

            dt1=dt1,
            nsteps1=nsteps1,
            dt2=dt2,
            nsteps2=nsteps2,
            evolve_seed=evolve_seed,

            ensemble=ensemble,
            pressure=pressure,
            tauS=tauS,
            pressure_couple=pressure_couple,
            barostat_gamma=barostat_gamma,

            outer_mask_diameter_fraction=(
                outer_mask_fraction
            ),
            nph_mask_controls_box=(
                nph_mask_controls_box
            ),
            nph_box_volume_ratio_bounds=(
                volume_bounds
            ),
            nph_safety_check_period=(
                safety_period
            ),

            random_location=random_location,
            base_folder=base_folder,

            # Overwrite only the evolution.
            overwrite=True,
            overwrite_initial=False,
            overwrite_source=False,

            # Require the original saved source.
            create_source_if_missing=False,
            reject_phase_separated_source=True,

            source_log_period=1_000,
            log_period=1_000,
            trajectory_period=1_000,
        )

        record["rerun_status"] = "returned_normally"
        record["result_status"] = result.get(
            "status",
            "",
        )
        record["outcome"] = result.get(
            "outcome",
            "",
        )

        progress_print(
            f"Finished rerun {rerun_number}/{total_reruns}; "
            f"status={record['result_status']}; "
            f"outcome={record['outcome'] or 'not_recorded'}"
        )

    except Exception as error:
        record["rerun_status"] = "raised_exception"
        record["error"] = (
            f"{type(error).__name__}: {error}"
        )

        progress_print(
            f"Rerun {rerun_number}/{total_reruns} failed: "
            f"{record['error']}"
        )
        progress_print(
            "Continuing to the next set of conditions."
        )

    rerun_results.append(record)

rerun_report = pd.DataFrame(
    rerun_results
)

progress_print("")
progress_print("All requested reruns have been attempted.")

display(rerun_report)
```
