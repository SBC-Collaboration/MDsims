```python
import pandas as pd
import matplotlib.pyplot as plt

from md_Helpers import (
    ProjectPaths,
    SQLiteRunDatabase,
    display_cavitation_table,
    fit_argon_lj_scale,
    plot_cavitation_seitz,
    query_argon_scaling_states,
)

paths = ProjectPaths()
database = SQLiteRunDatabase(paths.database)


# ============================================================
# 1. Select phase-separated states for the Argon scale fit
#
# These defaults reproduce the selection in Argon_Like.ipynb:
#   N_Cells = 40
#   Cumulative_LJ_Time = 200
#   Phase separated
#   rho_gas_unc <= 0.002
#   Therm_kT >= 0.85
#   Fixed-density runs only
# ============================================================

argon_states = query_argon_scaling_states(database)


# ============================================================
# 2. Fit epsilon/kB and sigma
#
# The Argon atomic mass, 39.948 u, is assigned automatically.
# ============================================================

argon_fit = fit_argon_lj_scale(argon_states)
argon_scale = argon_fit.scale

print("Argon-like LJ unit conversion:")
display(pd.Series(argon_scale.summary(), name="Value").to_frame())

print(
    "Temperature-ratio reduced chi-square:",
    argon_fit.temperature_reduced_chi2,
)

print(
    "Absolute-density reduced chi-square:",
    argon_fit.density_reduced_chi2,
)


# ============================================================
# 3. Select cavitation runs
# ============================================================

matches = display_cavitation_table(
    database,
    Nsteps=[500_000],
    Phase_Separation_Status="Separated",
    # N_Cells=45,
    # Mask_Radius=(2.0, 5.0),
    # Random_Location=0,
)


# ============================================================
# 4. Calculate Seitz Q and plot in physical Argon-like units
# ============================================================

figure, axis, seitz_results = plot_cavitation_seitz(
    matches,
    database,

    # Homogeneous states defining the EOS:
    eos_n_cells=45,
    eos_filters={
        "Nsteps": 200_000,
        "dt": 0.002,
        "Ensemble": "NVT",
        # "LJ_Mode": "xplor",
    },

    # Apply fitted Argon conversion:
    physical_scale=argon_scale,
    energy_unit="eV",       # Change to "keV", "J", or "kJ/mol"
    density_unit="mol/L",   # Change to "kg/m^3" or "1/nm^3"
)

plt.show()


# ============================================================
# 5. Inspect the physical Seitz values
# ============================================================

display(
    seitz_results[
        [
            "Run_ID",
            "Therm_kT",
            "rho_liquid",
            "rho_liquid_mol/L",
            "Q",
            "Q_uncertainty",
            "Q_eV",
            "Q_uncertainty_eV",
        ]
    ]
)
```
