```python
import matplotlib.pyplot as plt
from md_Helpers import open_run

run_ids = [
    "20260903204405",
    "20260903203452",
    "20260903202814",
]

fig, ax = plt.subplots(figsize=(9, 5))

for run_id in run_ids:
    run = open_run(run_id)
    logs = run.logs_dataframe()

    ax.plot(
        logs["this_lj_time"],
        logs["max_particle_speed"],
        label=run_id,
    )

ax.set_xlabel("LJ time within run")
ax.set_ylabel("Pressure")
ax.set_title("Pressure comparison")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```
