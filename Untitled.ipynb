```python
import numpy
import matplotlib.pyplot as plt

# -----------------------------
# Parameters
# -----------------------------
epsilon = 1.0
sigma = 1.0
r_cut = 2.5
r_on = 2.0   # must be < r_cut for xplor

r = numpy.linspace(0.95, 3.0, 1000)

# -----------------------------
# Base LJ potential
# -----------------------------
V_lj = 4 * epsilon * ((sigma / r)**12 - (sigma / r)**6)

# -----------------------------
# Shift mode
# -----------------------------
V_cut = 4 * epsilon * ((sigma / r_cut)**12 - (sigma / r_cut)**6)
V_shift = numpy.where(r < r_cut, V_lj - V_cut, 0.0)

# -----------------------------
# XPLOR switching function
# -----------------------------
S = numpy.ones_like(r)

middle = (r >= r_on) & (r <= r_cut)
outside = r > r_cut

S[middle] = (
    ((r_cut**2 - r[middle]**2)**2) *
    (r_cut**2 + 2 * r[middle]**2 - 3 * r_on**2)
) / (r_cut**2 - r_on**2)**3

S[outside] = 0.0

V_xplor = numpy.where(r < r_on, V_lj, S * V_lj)
V_xplor = numpy.where(r > r_cut, 0.0, V_xplor)

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(8, 5))
plt.plot(r, V_lj, label="LJ (unmodified)")
plt.plot(r, V_shift, label='mode = "shift"')
plt.plot(r, V_xplor, label='mode = "xplor"')

plt.axvline(r_on, linestyle="--", linewidth=1, label=f"r_on = {r_on}")
plt.axvline(r_cut, linestyle="--", linewidth=1, label=f"r_cut = {r_cut}")
plt.axhline(0, linestyle="--", linewidth=1)

plt.xlim(0.95, 3.0)
plt.ylim(-1.3, 1)
plt.xlabel("r")
plt.ylabel("V(r)")
plt.title("Lennard-Jones: none vs shift vs xplor")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
